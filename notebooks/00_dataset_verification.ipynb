```python
import os
import pickle
import numpy as np
import pandas as pd
import wfdb

DATA_DIR = os.path.join("..", "data")
print("Environment ready!")
```

```python
ptb_dir = os.path.join(DATA_DIR, "ptb-xl")
os.makedirs(ptb_dir, exist_ok=True)

print("Fetching PTB-XL sample...")
wfdb.dl_database('ptb-xl', ptb_dir, records=['records100/00000/00001_lr'])

ptb_rec = wfdb.rdrecord(os.path.join(ptb_dir, 'records100', '00000', '00001_lr'))

print("--- 1. PTB-XL ---")
print("Signal Shape:", ptb_rec.p_signal.shape)
print("Sampling Rate:", ptb_rec.fs, "Hz")
print("Channels:", ptb_rec.sig_name)
```

```python
mit_dir = os.path.join(DATA_DIR, "mit-bih")
os.makedirs(mit_dir, exist_ok=True)

print("Fetching MIT-BIH sample...")
wfdb.dl_database('mitdb', mit_dir, records=['100'])

mit_rec = wfdb.rdrecord(os.path.join(mit_dir, '100'))
mit_ann = wfdb.rdann(os.path.join(mit_dir, '100'), 'atr')

print("--- 2. MIT-BIH Arrhythmia ---")
print("Signal Shape:", mit_rec.p_signal.shape)
print("Sampling Rate:", mit_rec.fs, "Hz")
print("Channels:", mit_rec.sig_name)
print("Annotations Count:", len(mit_ann.sample))
```

```python
wesad_file = os.path.join(DATA_DIR, "wesad", "S2", "S2.pkl")

print("--- 3. WESAD ---")
if os.path.exists(wesad_file):
    with open(wesad_file, 'rb') as f:
        wesad_data = pickle.load(f, encoding='latin1')
    print("Keys:", list(wesad_data.keys()))
    print("Chest ECG Shape:", wesad_data['signal']['chest']['ECG'].shape, "(700 Hz)")
    print("Wrist PPG Shape:", wesad_data['signal']['wrist']['BVP'].shape, "(64 Hz)")
else:
    print(f"Place S2.pkl inside: {wesad_file}")
```

```python
dalia_file = os.path.join(DATA_DIR, "ppg-dalia", "S1", "S1.pkl")

print("--- 4. PPG-DaLiA ---")
if os.path.exists(dalia_file):
    with open(dalia_file, 'rb') as f:
        dalia_data = pickle.load(f, encoding='latin1')
    print("Keys:", list(dalia_data.keys()))
    print("Wrist PPG Shape:", dalia_data['signal']['wrist']['BVP'].shape)
    print("Chest ECG Shape:", dalia_data['signal']['chest']['ECG'].shape)
else:
    print(f"Place S1.pkl inside: {dalia_file}")
```

```python
multisite_dir = os.path.join(DATA_DIR, "multisite-ppg")

print("--- 5. Multi-site PPG (2026) ---")
if os.path.exists(multisite_dir) and len(os.listdir(multisite_dir)) > 0:
    files = os.listdir(multisite_dir)
    print("Files found:", files)
    sample_file = os.path.join(multisite_dir, files[0])
    if sample_file.endswith('.csv'):
        df = pd.read_csv(sample_file)
        print("Shape:", df.shape)
        print(df.head(2))
else:
    print(f"Directory empty: {multisite_dir}")
```
